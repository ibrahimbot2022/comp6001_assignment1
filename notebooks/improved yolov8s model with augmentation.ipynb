Average confidence

```python
def get_confidence_stats(results):
    confidences = []
    detections_per_image = []

    for r in results:
        num_boxes = len(r.boxes)
        detections_per_image.append(num_boxes)

        if num_boxes > 0:
            confs = r.boxes.conf.cpu().numpy()
            confidences.extend(confs.tolist())

    avg_conf = sum(confidences) / len(confidences) if confidences else 0
    return avg_conf, detections_per_image, confidences

blurred_avg_conf, blurred_per_image, blurred_confs = get_confidence_stats(blurred_results)
deblurred_avg_conf, deblurred_per_image, deblurred_confs = get_confidence_stats(deblurred_results)
sharp_avg_conf, sharp_per_image, sharp_confs = get_confidence_stats(sharp_results)

print("Blurred average confidence:", blurred_avg_conf)
print("Deblurred average confidence:", deblurred_avg_conf)
print("Sharp average confidence:", sharp_avg_conf)
```

random split of images ino train and test folders

```python
import random

image_files = sorted(os.listdir(deblurred_path))
image_files = [f for f in image_files if f.endswith(".png") or f.endswith(".jpg") or f.endswith(".jpeg")]

random.seed(42)
random.shuffle(image_files)

split_index = int(0.8 * len(image_files))
train_files = image_files[:split_index]
val_files = image_files[split_index:]

print("Train files:", len(train_files))
print("Val files:", len(val_files))
```

```python
import shutil

copied_train = 0
copied_val = 0
missing_labels = []

for filename in train_files:
    label_name = filename.rsplit(".", 1)[0] + ".txt"
    label_path = os.path.join(labels_source, label_name)

    if os.path.exists(label_path):
        shutil.copy(os.path.join(deblurred_path, filename), os.path.join(images_train, filename))
        shutil.copy(label_path, os.path.join(labels_train, label_name))
        copied_train += 1
    else:
        missing_labels.append(filename)

for filename in val_files:
    label_name = filename.rsplit(".", 1)[0] + ".txt"
    label_path = os.path.join(labels_source, label_name)

    if os.path.exists(label_path):
        shutil.copy(os.path.join(deblurred_path, filename), os.path.join(images_val, filename))
        shutil.copy(label_path, os.path.join(labels_val, label_name))
        copied_val += 1
    else:
        missing_labels.append(filename)

print("Copied train images:", copied_train)
print("Copied val images:", copied_val)
print("Missing labels:", len(missing_labels))
print("Examples missing labels:", missing_labels[:10])
```

```python
import os
import cv2
import numpy as np
from tqdm import tqdm
from scipy.signal import convolve2d

def motion_blur_kernel(size=15):
    kernel = np.zeros((size, size), dtype=np.float32)
    kernel[size // 2, :] = 1.0
    kernel /= kernel.sum()
    return kernel

def richardson_lucy(image, kernel, iterations=10):
    image = image.astype(np.float32) / 255.0
    estimate = np.full(image.shape, 0.5, dtype=np.float32)
    kernel_mirror = np.flipud(np.fliplr(kernel))

    for _ in range(iterations):
        conv_est = convolve2d(estimate, kernel, mode="same", boundary="symm")
        conv_est[conv_est == 0] = 1e-6
        relative_blur = image / conv_est
        estimate *= convolve2d(relative_blur, kernel_mirror, mode="same", boundary="symm")

    estimate = np.clip(estimate, 0, 1)
    return (estimate * 255).astype(np.uint8)

# clear folder just in case
for f in os.listdir(deblurred_path):
    os.remove(os.path.join(deblurred_path, f))

kernel = motion_blur_kernel(size=15)

for filename in tqdm(sorted(os.listdir(blurred_path))):
    img_gray = cv2.imread(os.path.join(blurred_path, filename), cv2.IMREAD_GRAYSCALE)
    restored = richardson_lucy(img_gray, kernel, iterations=10)
    cv2.imwrite(os.path.join(deblurred_path, filename), restored)

print("Deblurred files:", len(os.listdir(deblurred_path)))
```
